```python
def chunk_text(text, chunk_size=500, overlap=100):
    chunks = []
    start = 0

    while start < len(text):
        end = start + chunk_size
        chunk = text[start:end]
        chunks.append(chunk)
        start += chunk_size - overlap

    return chunks
```

```python
from pypdf import PdfReader

def extract_text_from_pdf(file_path):
    reader = PdfReader(file_path)
    text = ""

    for page in reader.pages:
        text += page.extract_text() + "\n"

    return text
```

```python
import re

def clean_text(text):
    text = re.sub(r'\n+', '\n', text)
    text = re.sub(r'\s+', ' ', text)
    return text.strip()
```

```python
import re

def normalize_text(text):
    # hapus spasi berlebih
    text = re.sub(r'\s+', ' ', text)

    # paksa tiap kalimat jadi baris baru
    text = re.sub(r'\.\s+', '.\n', text)

    return text.strip()
```

```python
def smart_chunk_v3(text, max_words=150):
    sentences = text.split("\n")
    chunks = []
    current_chunk = ""

    for sentence in sentences:
        if len((current_chunk + sentence).split()) <= max_words:
            current_chunk += " " + sentence
        else:
            chunks.append(current_chunk.strip())
            current_chunk = sentence

    if current_chunk:
        chunks.append(current_chunk.strip())

    return chunks
```

```python
def embed_passage(text):
    return model.encode(f"passage: {text}").tolist()
```

```python
def ingest_pdf(file_path, source_name):
    text = extract_text_from_pdf(file_path)
    text = clean_text(text)
    chunks = smart_chunk(text)

    for i, chunk in enumerate(chunks):
        embedding = embed_passage(chunk)

        supabase.table("document_e5large").insert({
            "content": chunk,
            "embedding": embedding,
            "metadata": {
                "kategori": "kampus"
            },
            "source": source_name,
            "chunk_index": i
        }).execute()

    print(f"✅ {len(chunks)} chunks berhasil di-insert!")
```

```python
from sentence_transformers import SentenceTransformer
from supabase import create_client
from groq import Groq
import os

# === CONFIG ===
SUPABASE_URL = "https://kwbgskhdkwpdkjwwjlpx.supabase.co"
SUPABASE_KEY = "YOUR_SUPABASE_KEY"
GROQ_API_KEY = os.environ["GROQ_API_KEY"]

# === INIT ===
model = SentenceTransformer("intfloat/multilingual-e5-large")
supabase = create_client(SUPABASE_URL, SUPABASE_KEY)
client = Groq(api_key=GROQ_API_KEY)

# === FUNCTIONS ===
def embed_query(text):
    return model.encode(f"query: {text}").tolist()

def search_similar(query_embedding, top_k=3):
    result = supabase.rpc("match_documents1", {
        "query_embedding": query_embedding,
        "match_count": top_k
    }).execute()
    return result.data

def build_context(results):
    return "\n\n".join([r["content"] for r in results])

def ask_llm(context, question):
    prompt = f"""
Jawab berdasarkan konteks berikut:
{context}

Pertanyaan:
{question}
"""
    response = client.chat.completions.create(
        model="openai/gpt-oss-120b",
        messages=[{"role": "user", "content": prompt}],
        temperature=0.2
    )
    return response.choices[0].message.content

def chatbot(question):
    query_embedding = embed_query(question)
    results = search_similar(query_embedding)
    context = build_context(results)

    print("=== CONTEXT ===")
    print(context)

    return ask_llm(context, question)
```

### RAG System Evaluation Metrics

To evaluate the performance of our Retrieval Augmented Generation (RAG) system, we will calculate the following metrics:

1.  **Accuracy**: Measures how often the LLM's answer is correct when compared to a ground truth. For this demonstration, we'll use a simplified check against expected answers.
2.  **Average Response Time (Latency)**: The average time taken for the `chatbot` to process a query and return a response.
3.  **Retrieval Relevance**: Assesses how well the retrieved documents (context) relate to the user's query. In this example, we'll define a simple relevance score based on keywords.


```python
import time

def evaluate_rag_system(test_cases):
    total_response_time = 0
    correct_answers_count = 0
    total_retrieval_relevance_score = 0

    print("\n--- Starting RAG System Evaluation ---")

    for i, case in enumerate(test_cases):
        question = case["question"]
        ground_truth = case["ground_truth"]
        expected_keywords = case["expected_relevant_keywords"]

        print(f"\n--- Test Case {i+1}: {question} ---")

        # Simulate chatbot process to capture intermediate results for evaluation
        query_embedding = embed_query(question)
        retrieved_docs = search_similar(query_embedding)
        context = build_context(retrieved_docs)

        start_time = time.time()
        llm_answer = ask_llm(context, question)
        end_time = time.time()
        response_time = end_time - start_time
        total_response_time += response_time

        print(f"LLM Answer: {llm_answer}")
        print(f"Response Time: {response_time:.2f} s")

        # --- Evaluate Accuracy (Simplified) ---
        # In a real scenario, this would involve more sophisticated NLP techniques
        # or human evaluation. Here, we do a simple keyword check.
        if all(keyword.lower() in llm_answer.lower() for keyword in ground_truth.split()):
            correct_answers_count += 1
            print("Accuracy: ✅ (Likely Correct)")
        else:
            print("Accuracy: ❌ (Potentially Incorrect)")

        # --- Evaluate Retrieval Relevance (Simplified) ---
        # Check if expected keywords are present in the retrieved context
        relevance_score = 0
        for keyword in expected_keywords:
            if keyword.lower() in context.lower():
                relevance_score += 1

        # Normalize relevance score
        normalized_relevance = relevance_score / len(expected_keywords) if expected_keywords else 0
        total_retrieval_relevance_score += normalized_relevance
        print(f"Retrieval Relevance Score: {normalized_relevance:.2f}/1")

    # --- Calculate Averages ---
    num_test_cases = len(test_cases)
    avg_response_time = total_response_time / num_test_cases
    accuracy_percentage = (correct_answers_count / num_test_cases) * 100
    avg_retrieval_relevance = (total_retrieval_relevance_score / num_test_cases) * 5 # Scale to 5-point scale

    print("\n--- Evaluation Summary ---")
    print(f"Accuracy = {accuracy_percentage:.2f}%")
    print(f"Average Response Time = {avg_response_time:.2f} s")
    print(f"Retrieval Relevance = {avg_retrieval_relevance:.2f}/5")
    print(f"Latency = {avg_response_time * 1000:.0f} ms")

# --- Define Test Cases ---
# For a real evaluation, you'd have a larger, diverse dataset with detailed ground truths.
test_cases = [
    {
        "question": "kapan pembayaran UKT?",
        "ground_truth": "tepat waktu sesuai jadwal akademik",
        "expected_relevant_keywords": ["UKT", "jadwal akademik", "administrasi"]
    },
    {
        "question": "dosen saya tidak merespon",
        "ground_truth": "hubungi 3 hari lalu mediasi prodi",
        "expected_relevant_keywords": ["dosen", "merespon", "mediasi", "prodi"]
    }
]

# --- Run Evaluation ---
evaluate_rag_system(test_cases)
```

```python
from sentence_transformers import SentenceTransformer
import torch
import time
import gc

# ========================
# INFO GPU
# ========================
print("GPU:", torch.cuda.get_device_name(0))
print("VRAM:", torch.cuda.get_device_properties(0).total_memory / 1e9, "GB")

# ========================
# LOAD MODEL
# ========================
model = SentenceTransformer("intfloat/e5-large", device="cuda")

# optional: hemat VRAM + lebih cepat
model = model.half()

# ========================
# DATA (besar biar realistis)
# ========================
texts = [
    "query: apa itu machine learning?",
    "passage: machine learning adalah cabang AI yang mempelajari pola dari data"
] * 10000   # = 20.000 teks

print("Total texts:", len(texts))

# ========================
# BENCHMARK FUNCTION
# ========================
def benchmark(batch_size):
    try:
        torch.cuda.empty_cache()
        gc.collect()

        start = time.time()

        with torch.no_grad():
            embeddings = model.encode(
                texts,
                batch_size=batch_size,
                show_progress_bar=True,
                convert_to_tensor=True
            )

        end = time.time()

        total_time = end - start
        throughput = len(texts) / total_time

        print(f"\n=== BATCH {batch_size} ===")
        print("Shape:", embeddings.shape)
        print("Time:", total_time)
        print("Throughput:", throughput, "texts/sec")

        return total_time, throughput

    except RuntimeError as e:
        print(f"\n❌ BATCH {batch_size} FAILED:", e)
        return None, None


# ========================
# RUN TEST
# ========================
results = {}

for bs in [64, 128, 256, 512, 1024]:
    t, th = benchmark(bs)
    if t is None:
        break
    results[bs] = (t, th)


# ========================
# SUMMARY
# ========================
print("\n=== SUMMARY ===")
for bs, (t, th) in results.items():
    print(f"Batch {bs}: {t:.3f}s | {th:.2f} texts/sec")

best = max(results.items(), key=lambda x: x[1][1])
print(f"\n🔥 BEST BATCH: {best[0]} (Throughput: {best[1][1]:.2f} texts/sec)")
```
